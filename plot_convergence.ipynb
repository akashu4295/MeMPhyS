```python
import pandas as pd
import matplotlib.pyplot as plt

# Base filename pattern
base_name = "Convergence_Solution_implicit_poly_3_Re_10_n_{}.csv"

# Range of files (n = 0 to 5)
n_values = range(6)

plt.figure(figsize=(8, 6))

dt = [0.082938, 0.024433, 0.007526, 0.004778, 0.002555, 0.001211]  # Example time step sizes for each n

for n in n_values:
    file_name = base_name.format(n)
    
    try:
        # Read CSV (assuming no header; adjust if needed)
        df = pd.read_csv(file_name, header=None)
        
        # Column 1 -> index 0 (time step)
        # Column 2 -> index 1 (steady state error)
        time_step = df.iloc[:, 0]
        error = df.iloc[:, 1]
        
        # Plot
        plt.plot(dt[n] * time_step, error, label=f"mesh {n+1}")
        print(f"time step size for mesh {n+1}: {dt[n]}")
    except FileNotFoundError:
        print(f"File not found: {file_name}")

# Labels and formatting
plt.xlabel("Time (s)")
plt.ylabel("Steady State Error")
plt.title("Convergence Plot")
plt.legend()
plt.grid()
# plt.xscale('log')
plt.yscale('log')
plt.ylim(1e-6, 1e-1)  # Adjust y-axis limits for better visualization
plt.xlim(0, 10)  # Limit x-axis to the first 2000 time steps
plt.tight_layout()
plt.show()
```
